## Obstacles and Cleaning
While preparing the dataset for import into PostgreSQL, I encountered several issues that required additional cleaning in Jupyter Notebook. The main obstacle was that the original Excel sheets contained text fields with commas, quotation marks, and other punctuation that caused the CSV files to break during import. PostgreSQL interpreted these characters as column separators or incomplete quoted fields, resulting in errors such as “unterminated CSV quoted field.”

To resolve this, I returned to Jupyter Notebook and performed several cleaning steps:

1. Removed commas from location and other text columns to prevent PostgreSQL from splitting fields incorrectly.

2. Stripped stray quotation marks that Excel had inserted around certain fields.

3. Removed inconsistent punctuation from descriptive columns that caused rows to be parsed incorrectly.

4. Converted all text fields to plain strings to ensure consistent formatting before export.

5. Re‑exported the cleaned data as UTF‑8 CSV files, which allowed PostgreSQL to import all three tables without errors.

## What advantage‑related events did each castaway participate in, and when did they occur?

```python
SELECT 
    c.full_name,
    c.season,
    a.event,
    a.day,
    a.episode
FROM castaways c
JOIN advantage_movement a
ON c.castaway_id = a.castaway_id
ORDER BY c.full_name, a.day;
```


![Import Screenshot](screenshots/join.png)

## How many advantages did the average winner have?

```python
SELECT 
    AVG(adv_count) AS avg_advantages_per_winner
FROM (
    SELECT 
        c.castaway_id,
        COUNT(a.advantage_id) AS adv_count
    FROM castaways c
    LEFT JOIN advantage_movement a
        ON c.castaway_id = a.castaway_id
    WHERE c.winner = TRUE
    GROUP BY c.castaway_id
) AS winner_advantages;
```


![Import Screenshot](screenshots/average.png)

## For each season, which castaway won and how many advantage events did that winner participate in?

```python
SELECT
    c.full_name AS winner_name,
    c.season,
    COUNT(a.advantage_id) AS total_advantages
FROM castaways c
LEFT JOIN advantage_movement a
    ON c.castaway_id = a.castaway_id
WHERE c.winner = TRUE
GROUP BY c.full_name, c.season
ORDER BY c.season;
```

## Survivor Insights

After completing my JOIN and GROUP BY analysis, I was able to quantify how often winners interacted with advantages during their seasons. By linking the castaways table to the advantage_movement table through castaway_id, I calculated the total number of advantage events each winner participated in and then averaged those values across all winners in the dataset. This revealed a clear gameplay trend: the average Survivor winner participated in approximately 2.4 advantage events, showing that while advantages are not required to win, they play a meaningful role in the strategies of many champions.
